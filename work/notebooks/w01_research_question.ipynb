## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

*30,000 Total Pages: The dataset gives us 30,000 published pages to evaluate across the site.

*54.2% In Active Decline: 16,262 out of 30,000 pages are currently losing search traffic (trend_direction == 'down').

*9,961 High-Impact Targets: Nearly 10,000 decaying pages generated $\ge 500$ impressions in the last 90 days, showing massive potential traffic to recover if refreshed.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Calculate real numbers from the starter dataset (df loaded in Section 1)
total_pages = len(df)
declining_pages = (df["trend_direction"].str.lower() == "down").sum()
declining_pct = declining_pages / total_pages
high_impact_declining = len(
    df[
        (df["trend_direction"].str.lower() == "down")
        & (df["impressions_90d"] >= 500)
    ]
)

print(f"1. Total pages in dataset: {total_pages:,}")
print(
    f"2. Pages in active decline: {declining_pages:,} ({declining_pct:.1%})"
)
print(
    f"3. High-impact decaying pages (>=500 impressions): {high_impact_declining:,}"
)
```
